```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from pyvolleydata.get_data import load_pbp

pd.set_option('display.max_columns', None)
```

```python
df = pd.DataFrame()
for yr in range(2024, 2027):
    temp = load_pbp('mlv', yr)
    df = pd.concat([df, temp])

df = df.set_index(['match_id', 'set', 'point_number'])
```

```python
df['contact_num'] = df.groupby(level=['match_id', 'set', 'point_number']).cumcount() + 1
df
```

```python
sns.countplot(data=df, x='outcome', hue='action')
```

```python
df[((df['action'] == 'D') & (df['outcome'] == '!'))]
```

```python
passes = (df["action"] == "R") | (df['action'] == "D")
df["pass_outcome"] = df["outcome"].where(passes)
df["pass_team"] = df["team_involved"].where(passes)

df["pass_outcome"] = (
    df.groupby(["match_id", "set_number", "point_number"])["pass_outcome"]
      .ffill()
)

df["pass_team"] = (
    df.groupby(["match_id", "set_number", "point_number"])["pass_team"]
      .ffill()
)

spikes = df["action"] == "Spike"

analysis = df.loc[
    spikes &
    (df["team_involved"] == df["pass_team"])
].copy()
```
