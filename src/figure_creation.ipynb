```python
# contingency counts + row totals for context
counts = contingency.copy()
counts["Total"] = counts.sum(axis=1)

# flag sparse cells (n < 30) for asterisking
SPARSE_THRESHOLD = 30
sparse_mask = contingency < SPARSE_THRESHOLD

# build display version with asterisks on sparse cells
display_df = contingency.astype(str)
col_names = ['#', '+', '!', '-', '=', '/', 'Total']
for col in contingency.columns:
    display_df[col] = contingency[col].astype(str) + sparse_mask[col].map(lambda x: " *" if x else "")
display_df["Total"] = counts["Total"]
display_df.index = ['Digs', 'Freeball Passes', 'Serve Receives']
display_df = display_df[col_names].rename_axis(None, axis=1).reset_index(names=['Pass Type'])
display_df
```
